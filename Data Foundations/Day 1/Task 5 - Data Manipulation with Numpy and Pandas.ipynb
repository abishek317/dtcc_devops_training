```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("movie_ticket_bookings.csv")
```

```python
print("========== 1. DATAFRAME INSPECTION ==========")
print("\nFirst 5 rows:\n", df.head())
print("\nLast 5 rows:\n", df.tail())
```

```python
print("\nShape (rows, columns):", df.shape)
print("\nColumn names:\n", df.columns)
print("\nDataFrame information:")
df.info()
```

```python
print("\nDescriptive statistics:\n", df.describe())
print("\nNull values by column:\n", df.isnull().sum())
print("\nNull check for each cell (first 5 rows):\n", df.isnull().head())
```

```python
print("\nNon-null count per column:\n", df.count())
print("\nUnique values per column:\n", df.nunique())
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python
print("\n========== 2. NUMPY ARRAYS / SERIES ==========")
tickets = df["Tickets"]
print("\nTickets Series:\n", tickets.head(10))
print("Series sum (ignores missing values):", tickets.sum())
```

```python
print("Series minimum:", tickets.min())
print("Series maximum:", tickets.max())
print("Series average:", tickets.mean())
```

```python
ticket_array = df["Tickets"].to_numpy()
print("\nTickets NumPy array:\n", ticket_array)
```

```python
print("ndim:", np.ndim(ticket_array))
print("shape:", np.shape(ticket_array))
print("size:", np.size(ticket_array))
```

```python
print("Total tickets:", np.nansum(ticket_array))
print("Minimum tickets:", np.nanmin(ticket_array))
print("Maximum tickets:", np.nanmax(ticket_array))
print("Average tickets:", np.nanmean(ticket_array))
```

```python
daily_tickets = (
    df.groupby("Date", as_index=False)["Tickets"]
      .sum()
)
```

```python
daily_array = daily_tickets["Tickets"].to_numpy()
print("\nDaily ticket totals:\n", daily_tickets)
print("Daily totals array shape:", daily_array.shape)
print("Daily totals array ndim:", daily_array.ndim)
print("Daily totals array size:", daily_array.size)
```

```python
print("Sum of daily tickets:", np.nansum(daily_array))
print("Average daily tickets:", np.nanmean(daily_array))
print("Maximum daily tickets:", np.nanmax(daily_array))
print("Minimum daily tickets:", np.nanmin(daily_array))
```

```python
print("\n========== 3. MULTIPLY / INCREASE BY 10% ==========")
df["TicketPrice_10pct_increase"] = df["TicketPrice"] * 1.10
print(df[["Movie", "TicketPrice", "TicketPrice_10pct_increase"]].head())
print("\n10% of the first 10 ticket prices:")
print(df["TicketPrice"].head(10).to_numpy() * 0.10)
```

```python
print("\n========== 4. FILTERING BY CONDITION ==========")
print("\nBookings with more than 2 tickets:")
print(df[df["Tickets"] > 2][["BookingID", "City", "Movie", "Tickets"]])

print("\nBookings with rating >= 4.5:")
print(df[df["Rating"] >= 4.5][["BookingID", "Movie", "Rating"]])

print("\nAction movie bookings:")
print(df[df["Category"] == "Action"][["BookingID", "City", "Movie", "Tickets"]])
```

```python
print("\n========== 5. CREATE REVENUE COLUMN ==========")

df["Revenue"] = df["Tickets"] * df["TicketPrice"]
print(df[["BookingID", "City", "Movie", "Tickets", "TicketPrice", "Revenue"]].head(10))

print("\nRevenue statistics:\n", df["Revenue"].describe())
```

```python
print("Total revenue:", df["Revenue"].sum())
print("Average booking revenue:", df["Revenue"].mean())
print("Minimum booking revenue:", df["Revenue"].min())
print("Maximum booking revenue:", df["Revenue"].max())
```

```python
print("\nRevenue by category:\n",
      df.groupby("Category")["Revenue"].agg(["sum", "mean", "min", "max", "count"]))
print("\nRevenue by movie:\n",
      df.groupby("Movie")["Revenue"].agg(["sum", "mean", "count"]).sort_values("sum", ascending=False))
print("\nRevenue by city (missing city excluded by groupby):\n",
      df.groupby("City")["Revenue"].agg(["sum", "mean", "count"]).sort_values("sum", ascending=False))
print("\nRevenue by date:\n", df.groupby("Date")["Revenue"].sum())
```

```python
print("\n========== 6. OPTIONAL DATA CLEANING ==========")
print("Missing values before cleaning:\n", df.isnull().sum())

df_clean = df.drop_duplicates().copy()
print("Rows before removing duplicates:", len(df))
print("Rows after removing duplicates:", len(df_clean))

df_clean["Rating"] = df_clean["Rating"].fillna(df_clean["Rating"].median())
print("\nMissing ratings after median fill:", df_clean["Rating"].isnull().sum())

df_clean.to_csv("movie_ticket_bookings_processed.csv", index=False)
print("Saved cleaned data to movie_ticket_bookings_processed.csv")
```
