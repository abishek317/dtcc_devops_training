```python
df = df.replace(r"^\s*$", np.nan, regex=True)
df.isnull().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df.shape
```

```python
df["Product"] = df["Product"].str.strip()
df["Product"] = df["Product"].str.title()

df["Category"] = df["Category"].str.strip()
df["Category"] = df["Category"].str.title()

df["City"] = df["City"].str.strip()
df["City"] = df["City"].str.title()

df.head(10)
```

```python
print(df["Product"].unique())
print(df["Category"].unique())
print(df["City"].unique())
```

```python
df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)

df["OrderDate"].dtype
```

```python
df[df["OrderDate"] > "2026-10-10"]
```

```python
df["OrderID"] = pd.to_numeric(df["OrderID"], errors="coerce")
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")

df.info()
```

```python
df.isnull().sum()
```

```python
df["Quantity"] = df["Quantity"].fillna(0)
```

```python
df["UnitPrice"] = df["UnitPrice"].fillna(
    df["UnitPrice"].median()
)
```

```python
df["City"] = df["City"].fillna("Unknown")
```

```python
df = df.dropna(subset=["OrderDate"]).copy()
df.isnull().sum()
```

```python
Q1 = df["UnitPrice"].quantile(0.25)
Q3 = df["UnitPrice"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df["UnitPrice"] < lower_limit) |
    (df["UnitPrice"] > upper_limit)
]

outliers
```

```python
df["UnitPrice"].plot(
    kind="box",
    title="Unit Price Outliers"
)

plt.ylabel("Unit Price")
plt.show()
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
df.head()
```

```python
print("Total Revenue:", df["Revenue"].sum())
print("Average Revenue:", df["Revenue"].mean())
print("Maximum Revenue:", df["Revenue"].max())
print("Minimum Revenue:", df["Revenue"].min())
```

```python
def price_category_function(price):
    if price < 1000:
        return "Low"
    elif price < 20000:
        return "Medium"
    else:
        return "High"
```

```python
df["PriceCategory"] = df["UnitPrice"].apply(
    price_category_function
)
df.head(10)
```
