# Data Cleaning

```python
df.replace(r"^\s*$", np.nan, regex=True, inplace=True)

df.replace(["N/A", "not available"], np.nan, inplace=True)

df.isnull().sum()
```

```python
df = df.drop_duplicates()

df.duplicated().sum()
```

```python
df["Region"] = df["Region"].str.strip().str.title()
df["Product"] = df["Product"].str.strip().str.title()
df["Channel"] = df["Channel"].str.strip().str.title()

print(df["Region"].unique())
print(df["Product"].unique())
print(df["Channel"].unique())
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df[["OrderID", "Date"]]
```

```python
df["Date"].isnull().sum()
```

```python
df["Units"] = df["Units"].replace({"five": "5"})

df["DiscountPct"] = df["DiscountPct"].str.replace(
    "%", "", regex=False
)

df["DiscountPct"] = pd.to_numeric(
    df["DiscountPct"], errors="coerce"
)

df.loc[df["DiscountPct"] > 1, "DiscountPct"] /= 100

df["Units"] = pd.to_numeric(df["Units"], errors="coerce")
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")
```

```python
df.isnull().sum()
```

```python
df.loc[df["Units"] < 0, "Units"] = np.nan

df["Units"] = df["Units"].fillna(0)
df["DiscountPct"] = df["DiscountPct"].fillna(0)
df["UnitPrice"] = df["UnitPrice"].fillna(df["UnitPrice"].median())
```

```python
df.isnull().sum()
```

```python
df["Date"] = df["Date"].fillna(
    df["Date"].dropna().sort_values().iloc[len(df["Date"].dropna()) // 2]
)

df["Region"] = df["Region"].fillna("Unknown")
df["Product"] = df["Product"].fillna("Unknown")
df["Channel"] = df["Channel"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df.info()
```

# New Column for Revenue

```python
df["Revenue"] = (
    df["Units"] *
    df["UnitPrice"] *
    (1 - df["DiscountPct"])
)

df[["OrderID", "Product", "Units", "UnitPrice", "DiscountPct", "Revenue"]]
```

```python
# Grouping and Aggregation 
```

```python
product_summary = (
    df.groupby("Product")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        UnitsSold=("Units", "sum"),
        Orders=("OrderID", "count"),
        AverageRevenue=("Revenue", "mean")
    )
    .sort_values("TotalRevenue", ascending=False)
)

product_summary
```

```python
revenue_by_region = (
    df.groupby("Region")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

revenue_by_region
```

```python
region_summary = (
    df.groupby("Region")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        AverageRevenue=("Revenue", "mean"),
        Orders=("OrderID", "count"),
        UnitsSold=("Units", "sum")
    )
    .sort_values("TotalRevenue", ascending=False)
)

region_summary
```

```python
channel_summary = (
    df.groupby("Channel")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        Orders=("OrderID", "count"),
        UnitsSold=("Units", "sum"),
        AverageRevenue=("Revenue", "mean")
    )
    .sort_values("TotalRevenue", ascending=False)
)

channel_summary
```

```python
df["Month"] = df["Date"].dt.to_period("M").astype(str)

monthly_summary = (
    df.groupby("Month")
    .agg(
        TotalRevenue=("Revenue", "sum"),
        Orders=("OrderID", "count"),
        UnitsSold=("Units", "sum")
    )
    .sort_index()
)

monthly_summary
```
